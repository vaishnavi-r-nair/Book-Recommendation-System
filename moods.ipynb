```python
import pandas as pd
import numpy as np
from transformers import pipeline
```

```python
books_df = pd.read_csv("./dataset/books_dataset.csv")
books_df = books_df.fillna("No description")
books_df.head()
```

```python
classifier = pipeline("zero-shot-classification", model="facebook/bart-large-mnli", device="mps") # Load the model on 'mps' from Apple Silicon, use cuda on google collab
mood_labels = ["uplifting", "inspiring", "funny", "melancholic", "thrilling","romantic",]
```

```python
def classify_mood(description):
    result = classifier(description, mood_labels)
    return result['labels'][0]

books_df['mood'] = books_df['description'].apply(classify_mood)
```

```python
books_df.to_csv("./dataset/books_dataset_mood.csv")
```
